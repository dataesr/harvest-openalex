```python
def query_openalex(query, start_year, end_year, per_page):
    url = f'https://api.openalex.org/works?filter={query}'
    url += f',publication_year:{start_year}-{end_year}'
    url += f'&per_page={per_page}'
    url += f'&api_key={OPENALEX_API_KEY}'
    r = requests.get(url)
    return r.json()
```
