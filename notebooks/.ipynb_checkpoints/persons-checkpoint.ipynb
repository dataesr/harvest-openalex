```python
import os
import requests

from urllib.parse import quote_plus
import json

def get_aurehal_author():
    nb_rows = 10000
    cursor='*'
    data = []
    while True:
        url = f'https://api.archives-ouvertes.fr/ref/author/?q=*:*&wt=json&fl=*&sort=docid asc&rows={nb_rows}&cursorMark={cursor}'
        r = requests.get(url)
        res = r.json()
        new_cursor = quote_plus(res['nextCursorMark'])
        data += res['response']['docs']
        if new_cursor == cursor:
            break
        cursor = new_cursor
        break
    return data

data = get_aurehal_author()
```
